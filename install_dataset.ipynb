```python
import os
from dotenv import load_dotenv
from huggingface_hub import login, snapshot_download

# 1. โหลดข้อมูลตัวแปรสภาพแวดล้อมจากไฟล์ .env
load_dotenv()

# 2. ดึงค่า Token ออกมาจากตัวแปร HF_TOKEN
hf_token = os.getenv("HF_TOKEN")

# ตรวจสอบว่าดึง Token มาได้หรือไม่
if not hf_token:
    raise ValueError("ไม่พบ HF_TOKEN ในไฟล์ .env กรุณาตรวจสอบการตั้งค่า")

# 3. ยืนยันตัวตนผ่าน Token
login(token=hf_token)

# 4. ดาวน์โหลดโฟลเดอร์ลงเครื่อง
snapshot_download(
    repo_id="DefectSpectrum/Defect_Spectrum",
    repo_type="dataset",
    allow_patterns="DS-MVTec/*",
    local_dir="/Users/jaypoom/Desktop/AI Project/Dataset/MMAD",
    max_workers=4,
)
```

```python
import os

# Path ปลายทางที่คุณตั้งไว้
target_dir = "/Users/jaypoom/Desktop/AI Project/Dataset/MMAD/DS-MVTec"

total_files = 0
zero_byte_files = []

# เดินสำรวจไฟล์ทั้งหมดในโฟลเดอร์ย่อย
for root, dirs, files in os.walk(target_dir):
    for file in files:
        total_files += 1
        file_path = os.path.join(root, file)
        # เช็คว่ามีไฟล์ขนาด 0 Bytes (ดาวน์โหลดไม่สมบูรณ์) หรือไม่
        if os.path.getsize(file_path) == 0:
            zero_byte_files.append(file_path)

print(f"จำนวนไฟล์ทั้งหมดในโฟลเดอร์ DS-MVTec: {total_files} ไฟล์")

if zero_byte_files:
    print(f"⚠️ พบไฟล์ขนาด 0 Bytes จำนวน {len(zero_byte_files)} ไฟล์ (ดาวน์โหลดไม่สมบูรณ์):")
    for f in zero_byte_files[:5]:  # แสดง 5 รายการแรก
        print(f"  - {f}")
else:
    print("✅ ไฟล์ทั้งหมดสมบูรณ์และไม่มีไฟล์เสีย (0 Bytes) ครับ!")
```
